## 1. No Free Lunch Theorem (informal statement + implication)

Averaged over **all possible** data-generating distributions, every
learning algorithm has the same expected performance — no single algorithm
is universally best. This is why algorithm choice always depends on
assumptions about the specific problem (which is exactly the point of
`11_Algorithm_Selection_and_Comparison/01_how_to_choose_an_algorithm.ipynb`)
— there's no mathematically-justified default "best" algorithm independent
of the data.
