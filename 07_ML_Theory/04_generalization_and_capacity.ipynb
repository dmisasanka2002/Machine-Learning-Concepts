## 1. Generalization, Model Capacity, and VC Dimension

**Generalization gap**: $R(h) - \hat R(h)$ - how much worse a model does
on new data vs. training data. Learning theory bounds this gap in terms of
how "flexible" $\mathcal H$ is.

**VC (Vapnik-Chervonenkis) dimension** - formal definition: the VC
dimension of $\mathcal H$ is the largest number of points that $\mathcal H$
can **shatter** (label in every possible way using some $h\in\mathcal H$).

### Worked example: VC dimension of a linear classifier in 2D

Claim: $VC(\text{2D linear classifiers}) = 3$.
- **3 points can be shattered**: take 3 non-collinear points. There are
  $2^3=8$ possible labelings. For each, you can find a separating line
  (verify by drawing all 8 cases - 2 all-same-class trivial cases, 6
  cases each separable by a line since only 3 points can't force a
  non-linear pattern).
- **4 points cannot always be shattered**: take 4 points in "XOR"
  configuration (two classes at opposite corners of a small square: e.g.
  class A at $(0,0)$ and $(1,1)$, class B at $(0,1)$ and $(1,0)$). No
  single straight line separates A from B - this specific labeling is
  **not** achievable, so 4 points cannot always be shattered.
$$\Rightarrow VC(\text{2D linear classifiers}) = 3\ (=d+1 \text{ for } d\text{-dimensional linear classifiers})$$

**Why VC dimension matters**: (informal statement of the PAC bound)
$$R(h) \le \hat R(h) + O\!\left(\sqrt{\frac{VC(\mathcal H)\log n}{n}}\right)\quad \text{with high probability}$$
Higher VC dimension (more flexible model) → looser guarantee on
generalization for a fixed $n$ → need more data to trust the empirical
risk as a proxy for true risk. This is the formal version of "more complex
models need more data," connecting directly to Bias-Variance: high VC
dimension ↔ low bias, high variance.
