## 1. Hypothesis Space, Loss, Risk, and ERM

- **Hypothesis space** $\mathcal H$: the set of functions the learning
  algorithm is allowed to choose from (e.g., all linear functions, all
  depth-≤5 decision trees).
- **Loss function** $L(y,\hat y)$: cost of one prediction (squared error,
  0-1 loss, cross-entropy, ...).
- **True (expected) risk**: $R(h) = E_{(x,y)\sim \mathcal D}[L(y,h(x))]$ —
  the quantity we actually want to minimize, but can't compute directly
  since we don't know the true data distribution $\mathcal D$.
- **Empirical risk**: $\hat R(h) = \frac1n\sum_{i=1}^n L(y_i,h(x_i))$ — what
  we *can* compute, using the training sample.
- **Empirical Risk Minimization (ERM)**: $\hat h = \arg\min_{h\in\mathcal H}\hat R(h)$.

Every algorithm in this repo is an instance of ERM (or regularized ERM,
$\arg\min_h \hat R(h)+\lambda\,\Omega(h)$) over a specific $\mathcal H$ and
$L$: Linear Regression is ERM over linear functions with squared loss;
Logistic Regression is ERM over linear-in-logit functions with
cross-entropy loss; a Decision Tree is ERM over the space of tree-structured
partitions with (typically) 0-1 or Gini-related loss.

### Worked example: why minimizing empirical risk isn't automatically minimizing true risk

Suppose the true relationship is $y=2x+\text{noise}$, but by chance our
7-point training sample happens to have an outlier making the "best-fit by
empirical risk" line noticeably different from the population-optimal
line. Simulated: draw $n=7$ points from $y=2x+\mathcal N(0,1)$, plus one
extreme outlier from $y=2x+10$. The OLS line fit on this contaminated
sample (empirical risk minimizer) has a different slope than the OLS line
fit on 10,000 clean points from the same distribution (a good proxy for
the true-risk minimizer) - quantifying the **generalization gap**
$R(\hat h)-\hat R(\hat h)$ directly.
