# **Bayes Optimal Classifier and Bayes Error**

**Prerequisites:** No Free Lunch, Probability primer | **Next:** (end of `07_ML_Theory/` core sequence — proceed to `08_Classical_Reinforcement_Learning/` or `09_Interpretability/`) | **Depth tier:** Core

The **Bayes optimal classifier** for 0-1 loss is $h^*(x)=\arg\max_y P(y\mid x)$
— provably the best possible classifier for the *true* $P(y\mid x)$, with
error rate (**Bayes error**) $= E_x[1-\max_y P(y\mid x)]$. No classifier —
however powerful — can beat the Bayes error, because it reflects
irreducible overlap between classes' true conditional distributions
(directly parallels the irreducible-noise term $\sigma^2$ in the
bias-variance decomposition).

### Worked example
Suppose at a specific $x$, $P(y=1\mid x)=0.7$, $P(y=0\mid x)=0.3$. The
Bayes-optimal decision is to predict class 1 (higher posterior), with
local error rate $\min(0.7,0.3)=0.3$ — even the *perfect* classifier is
wrong 30% of the time at this $x$, because the classes genuinely overlap
there; no amount of additional data or model complexity fixes this.
