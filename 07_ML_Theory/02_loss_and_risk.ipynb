## 1. Theory

`01_hypothesis_spaces_and_erm.ipynb` introduced empirical risk as the
computable proxy for the true (expected) risk. This notebook focuses
specifically on that risk/empirical-risk distinction and why it's the
central unsolved gap every generalization argument in this repo (VC
dimension, bias-variance, cross-validation) is trying to bound or
estimate.

## 2. Mathematical Derivation - the definitions, isolated

$$\text{True risk: } R(h) = E_{(x,y)\sim\mathcal D}[L(y,h(x))] \qquad\text{(unobservable)}$$
$$\text{Empirical risk: } \hat R(h) = \frac1n\sum_{i=1}^nL(y_i,h(x_i)) \qquad\text{(computable)}$$
By the Law of Large Numbers (a direct consequence of the CLT machinery in
`00_Mathematics\03_probability.ipynb` & `00_Mathematics\04_statistics.ipynb`), $\hat R(h)\to R(h)$ as
$n\to\infty$ **for a FIXED $h$** - the subtlety every bound in this
section (VC dimension, `07_ML_Theory\06_vc_dimension_and_pac.ipynb`) exists to
handle is that ERM doesn't use a fixed $h$; it *searches* over
$\mathcal H$ to find the $\hat R$-minimizer, which can overfit to the
particular sample even as $\hat R(\hat h)\to R(\hat h)$ holds for each
individual candidate.

## 3. Worked Numerical Example

Flip a fair coin 5 times, record heads=1/tails=0: with seed 0 this gives
$\{1,1,1,0,0\}$ - empirical mean (empirical estimate of $P(\text{heads})$)
$=3/5=0.6$. True probability of heads is $0.5$ - the empirical estimate
from just 5 flips is a poor proxy for the true risk purely from
small-sample noise, before any overfitting-via-search enters at all -
worth isolating this base-rate noise from the *additional* problem of
searching over many candidate $h$.

```python
import numpy as np
rng = np.random.default_rng(0)
flips = rng.integers(0,2,size=5)
print(flips, flips.mean())   # small-n empirical mean, noisy around true 0.5

flips_large = rng.integers(0,2,size=100000)
print(flips_large.mean())    # converges close to 0.5, LLN in action
```
