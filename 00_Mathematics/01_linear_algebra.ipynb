## 1. Vectors and Basic Operations

A feature vector for one data point with $n$ features:
$$\mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{bmatrix} \in \mathbb{R}^n$$

**Dot product:**
$$\mathbf{a}\cdot\mathbf{b} = \sum_{i=1}^n a_i b_i = \|\mathbf{a}\|\,\|\mathbf{b}\|\cos\theta$$

**Euclidean (L2) norm:** $\|\mathbf{x}\|_2 = \sqrt{\sum_i x_i^2}$
**Manhattan (L1) norm:** $\|\mathbf{x}\|_1 = \sum_i |x_i|$

### Worked numerical example

Let $\mathbf{a}=(3,4)$, $\mathbf{b}=(1,2)$.
- $\mathbf{a}\cdot\mathbf{b} = 3(1)+4(2) = 11$
- $\|\mathbf{a}\|_2=\sqrt{3^2+4^2}=\sqrt{25}=5$
- $\|\mathbf{b}\|_2=\sqrt{1+4}=\sqrt5\approx2.236$
- $\cos\theta = \dfrac{11}{5\times2.236}=\dfrac{11}{11.18}\approx0.9838 \Rightarrow \theta\approx10.3^\circ$

This is exactly the "cosine similarity" used later in KNN/text classification.

## 2. Matrices and Matrix Multiplication

A design matrix stacks $m$ samples of $n$ features:
$$X = \begin{bmatrix} x_1^{(1)} & \cdots & x_n^{(1)} \\ \vdots & \ddots & \vdots \\ x_1^{(m)} & \cdots & x_n^{(m)} \end{bmatrix} \in \mathbb{R}^{m\times n}$$

Matrix multiplication $(AB)_{ij} = \sum_k A_{ik}B_{kj}$ requires inner dimensions to match.

### Worked example


$$A=\begin{bmatrix}1&2\\3&4\end{bmatrix},\quad B=\begin{bmatrix}5&6\\7&8\end{bmatrix}$$
$$AB=\begin{bmatrix}1(5)+2(7) & 1(6)+2(8)\\3(5)+4(7) & 3(6)+4(8)\end{bmatrix}=\begin{bmatrix}19&22\\43&50\end{bmatrix}$$

## 3. Transpose and Inverse

$(AB)^T = B^T A^T$. A square matrix $A$ is invertible iff $\det(A)\neq0$, and
$AA^{-1}=I$. For a $2\times2$ matrix:
$$A^{-1} = \frac{1}{ad-bc}\begin{bmatrix}d&-b\\-c&a\end{bmatrix},\quad A=\begin{bmatrix}a&b\\c&d\end{bmatrix}$$

### Worked example

$$A=\begin{bmatrix}4&7\\2&6\end{bmatrix},\ \det(A)=4(6)-7(2)=24-14=10$$
$$A^{-1}=\frac{1}{10}\begin{bmatrix}6&-7\\-2&4\end{bmatrix}=\begin{bmatrix}0.6&-0.7\\-0.2&0.4\end{bmatrix}$$
This is the exact operation used in the Linear Regression Normal Equation
$(X^TX)^{-1}$ (see `08-Linear-Regression-Additions.md`).

## 4. Eigenvalues and Eigenvectors

For square matrix $A$, $\mathbf{v}\neq0$ is an eigenvector with eigenvalue
$\lambda$ if:
$$A\mathbf{v} = \lambda\mathbf{v} \iff (A-\lambda I)\mathbf{v}=0$$
Nontrivial solutions require $\det(A-\lambda I)=0$ — the **characteristic
equation**.

### Worked example (used again, identically, in PCA)

$$A=\begin{bmatrix}2&1\\1&2\end{bmatrix}$$
Characteristic equation:
$$\det\begin{bmatrix}2-\lambda&1\\1&2-\lambda\end{bmatrix}=(2-\lambda)^2-1=0$$
$$\lambda^2-4\lambda+3=0 \Rightarrow (\lambda-1)(\lambda-3)=0 \Rightarrow \lambda_1=1,\ \lambda_2=3$$

For $\lambda_2=3$: $(A-3I)\mathbf{v}=0 \Rightarrow \begin{bmatrix}-1&1\\1&-1\end{bmatrix}\mathbf{v}=0 \Rightarrow v_1=v_2$.
Normalized eigenvector: $\mathbf{v}_2=\frac{1}{\sqrt2}(1,1)$.

For $\lambda_1=1$: similarly $v_1=-v_2 \Rightarrow \mathbf{v}_1=\frac{1}{\sqrt2}(1,-1)$.

Note $\mathbf{v}_1\cdot\mathbf{v}_2=0$ — eigenvectors of a **symmetric**
matrix (like a covariance matrix) are always orthogonal. This is exactly
why PCA's principal components are orthogonal (proved formally in
`14-PCA-Additions.md`).


## 5. Matrix Calculus (needed for every derivation in this repo)

$$\nabla_{\mathbf{w}}(\mathbf{w}^T\mathbf{a}) = \mathbf{a}, \qquad \nabla_{\mathbf{w}}(\mathbf{w}^TA\mathbf{w}) = (A+A^T)\mathbf{w} \xrightarrow{A\ \text{symmetric}} 2A\mathbf{w}$$

These two identities alone are all that's needed to derive the Normal
Equation for Linear Regression and the gradient of the Logistic Regression
cost function.

## 6. Python — quick verification of everything above

```python
import numpy as np

a = np.array([3, 4]); b = np.array([1, 2])
print("dot:", a @ b)                       # 11
print("cos theta:", (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b)))

A = np.array([[4, 7], [2, 6]])
print("inverse:\n", np.linalg.inv(A))

M = np.array([[2, 1], [1, 2]])
eigvals, eigvecs = np.linalg.eig(M)
print("eigenvalues:", eigvals)             # [1. 3.]
print("eigenvectors:\n", eigvecs)
```
