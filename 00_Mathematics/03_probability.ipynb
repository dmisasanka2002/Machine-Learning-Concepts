# **Basics Probability for Machine Learning**

## 1. Axioms and Conditional Probability

$$0\le P(A)\le1,\quad P(\Omega)=1,\quad P(A\cup B)=P(A)+P(B)-P(A\cap B)$$
$$P(A\mid B)=\frac{P(A\cap B)}{P(B)},\quad P(B)>0$$

## 2. Bayes' Theorem - full derivation

From the definition of conditional probability twice:
$$P(A\cap B)=P(A\mid B)P(B) = P(B\mid A)P(A)$$
$$\Rightarrow\ P(A\mid B)=\frac{P(B\mid A)P(A)}{P(B)}$$

In ML terms (used directly in Naive Bayes, `03_Supervised_Learning\Classification\07_naive_bayes.ipynb`):
$$P(y\mid \mathbf{x}) = \frac{P(\mathbf{x}\mid y)\,P(y)}{P(\mathbf{x})}
\quad\text{posterior} = \frac{\text{likelihood}\times\text{prior}}{\text{evidence}}$$

### Worked numerical example (medical test - classic Bayes' example)

A disease has prevalence $P(D)=0.01$. A test has:
- Sensitivity $P(+\mid D)=0.95$
- False-positive rate $P(+\mid \neg D)=0.05$

$$P(+) = P(+\mid D)P(D)+P(+\mid\neg D)P(\neg D) = 0.95(0.01)+0.05(0.99)=0.0095+0.0495=0.059$$
$$P(D\mid +) = \frac{0.95\times0.01}{0.059} = \frac{0.0095}{0.059} \approx 0.161$$

Even with a 95%-accurate test, a positive result means only a **16.1%**
chance of actually having the disease, because the disease is rare. This
exact phenomenon is why the repo's `cardio_modify.csv` / `kyphosis.csv`
(imbalanced medical data) need the imbalanced-data-handling notebook
(`02_Data\11_imbalanced_data_handling.ipynb`).

## 3. Random Variables and Key Distributions

**Bernoulli** (used in Logistic Regression's label $y\in\{0,1\}$):
$$P(Y=y)=p^y(1-p)^{1-y}$$

**Gaussian / Normal** (used in GMM, and assumed for residuals in Linear Regression):
$$f(x)=\frac{1}{\sigma\sqrt{2\pi}}\exp\!\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

**Multivariate Gaussian** (used directly in GMM):
$$f(\mathbf{x})=\frac{1}{(2\pi)^{n/2}|\Sigma|^{1/2}}\exp\!\left(-\tfrac12(\mathbf{x}-\mu)^T\Sigma^{-1}(\mathbf{x}-\mu)\right)$$
