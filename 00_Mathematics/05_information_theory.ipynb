## 1. Theory

Entropy, cross-entropy, and KL divergence already appeared *implicitly* in
this repo — Decision Trees' splitting criterion (`03_Supervised_Learning\Classification\11_decision_trees.ipynb`),
Logistic Regression's loss (`03_Supervised_Learning\Classification\01_logistic_regression.ipynb`), and
GMM's likelihood (`04_Unsupervised_Learning\08_gmm_and_em.ipynb`) — without ever being named as
"information theory" or generalized beyond their specific use. This
notebook makes the general tools explicit, since ML papers use this
vocabulary constantly and assume it as background.

## 2. Mathematical Derivation

**Entropy** (already derived for Decision Trees, restated generally):
$$H(P) = -\sum_x P(x)\log_2P(x)$$
Measures the average "surprise"/uncertainty in a distribution - maximized
when $P$ is uniform, zero when $P$ is a point mass.

**Cross-entropy** - the expected surprise of using distribution $Q$ to
"encode" outcomes actually drawn from $P$:
$$H(P,Q) = -\sum_x P(x)\log_2Q(x)$$
This IS exactly Logistic Regression's loss function, with $P$ = the true
label (a point mass at $y_i$) and $Q$ = the model's predicted probability
- worth stating explicitly as the same object, not a coincidental
naming similarity.

**KL Divergence** - how much cross-entropy exceeds entropy; a (non-symmetric)
"distance" from $Q$ to $P$:
$$D_{KL}(P\|Q) = H(P,Q) - H(P) = \sum_x P(x)\log_2\frac{P(x)}{Q(x)}$$
**Derivation that $D_{KL}\ge0$ always** (Gibbs' inequality, via Jensen's
inequality on the concave $\log$ function):
$$-D_{KL}(P\|Q) = \sum_xP(x)\log\frac{Q(x)}{P(x)} \le \log\sum_xP(x)\frac{Q(x)}{P(x)} = \log\sum_xQ(x) = \log1=0$$

so $D_{KL}(P\|Q)\ge0$, with equality iff $P=Q$ everywhere — this is
*why* minimizing cross-entropy (equivalently, KL divergence, since $H(P)$
is constant w.r.t. the model) is a sensible training objective: it's
minimized exactly when the model's distribution matches the true one.

**Mutual Information** - how much knowing $X$ reduces uncertainty about
$Y$:
$$I(X;Y) = H(Y) - H(Y\mid X) = D_{KL}\big(P(X,Y)\,\|\,P(X)P(Y)\big)$$
Zero iff $X,Y$ independent - a general, nonlinear alternative to
correlation for measuring dependence (correlation only captures *linear*
dependence; mutual information captures any dependence).

## 3. Worked Numerical Example

True label distribution $P=(1,0)$ (definitely class 1), model prediction
$Q=(0.8,0.2)$ (80% confident in class 1).
$$H(P) = -1\log_21 - 0\log_20 = 0 \text{ (a point mass has zero entropy)}$$
$$H(P,Q) = -1\log_2(0.8) - 0\log_2(0.2) = -\log_2(0.8) = 0.322 \text{ bits}$$
$$D_{KL}(P\|Q) = 0.322 - 0 = 0.322 \text{ bits}$$
This 0.322 is exactly the cross-entropy loss value Logistic Regression
would report for this single prediction (in bits — natural-log/"nats" is
the more common convention in ML libraries, off by a constant factor of
$\ln2$).
