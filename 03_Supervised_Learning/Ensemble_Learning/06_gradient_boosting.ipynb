## Gradient Boosting - full derivation (functional gradient descent)

Fit $F_0(x)=\arg\min_\gamma \sum_i L(y_i,\gamma)$ (e.g. the mean, for
squared-error loss). At each stage $t$, compute the **pseudo-residuals**
(negative gradient of the loss w.r.t. the current prediction):
$$r_{it} = -\left[\frac{\partial L(y_i,F(x_i))}{\partial F(x_i)}\right]_{F=F_{t-1}}$$
For squared-error loss $L=\frac12(y-F)^2$: $r_{it}=y_i-F_{t-1}(x_i)$ - i.e.
**literally the residual**, which is why "fit a tree to the residuals" is
the common description. Fit a new weak learner $h_t$ to predict $r_{it}$,
then update $F_t(x) = F_{t-1}(x) + \nu\, h_t(x)$, with learning rate
$\nu\in(0,1]$ (shrinkage) trading off more rounds for better generalization.

### Worked numerical example (squared-error loss, 4 points, 2 rounds)

Data: $x=\{1,2,3,4\}$, $y=\{2.0,2.5,4.5,5.0\}$.
$F_0(x) = \bar y = 3.5$ (constant model minimizing squared error).
Residuals round 1: $r = y-F_0 = \{-1.5,-1.0,1.0,1.5\}$.
Suppose the weak learner (a depth-1 stump splitting at $x\le2.5$) predicts
the average residual in each region: left region $\{-1.5,-1.0\}$→mean
$-1.25$; right region $\{1.0,1.5\}$→mean $1.25$. With $\nu=0.5$:
$$F_1(x) = F_0(x) + 0.5\, h_1(x) = \{3.5-0.625,\ 3.5-0.625,\ 3.5+0.625,\ 3.5+0.625\} = \{2.875, 2.875, 4.125, 4.125\}$$
New residuals: $y-F_1=\{-0.875,-0.375,0.375,0.875\}$ - smaller in magnitude
than round 1's residuals ($\{-1.5,-1,1,1.5\}$), confirming the model is
converging. XGBoost/LightGBM (covered at practical level in
`07_xgboost_lightgbm_catboost.ipynb`) extend this with second-order
(Newton) gradient info and explicit regularization on tree complexity -
worth *stating* the extended objective but not re-deriving from scratch at
this course level.
