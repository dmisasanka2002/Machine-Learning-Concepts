# **XGBoost Lightgbm CatBoost**

- **XGBoost**: adds an explicit regularization term on tree complexity to
  the objective (penalizing both the number of leaves and the magnitude of
  leaf weights - a Ridge-style $\ell_2$ penalty on leaf values, same
  mechanism as `03_Supervised_Learning\Regression\04_ridge_regression.ipynb` §2, applied to tree leaf weights
  instead of linear-model coefficients), and uses **second-order**
  (Newton) information - both the gradient AND Hessian of the loss - for
  a more accurate per-step correction than Gradient Boosting's
  gradient-only pseudo-residuals.

- **LightGBM**: same underlying gradient boosting math, but uses a
  histogram-based approach to bucket continuous features before searching
  splits (turning the $O(nd\log n)$-per-level split search into
  $O(n\cdot\text{bins})$, a real complexity win for large $n$), and grows
  trees leaf-wise (best-first) rather than level-wise - faster but more
  prone to overfitting on small datasets, a genuine bias-variance
  trade-off decision, not a strict improvement.

- **CatBoost**: same core boosting math, specialized handling of
  categorical features (avoiding naive one-hot expansion via an
  ordered-statistics target-encoding scheme designed to reduce a specific
  leakage risk - predicting a category's encoding using only *previously
  seen* rows in a random permutation, preventing the target-leakage that
  naive target encoding would otherwise introduce).
