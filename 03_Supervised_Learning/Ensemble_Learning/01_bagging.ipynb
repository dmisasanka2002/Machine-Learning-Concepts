## 1. Bagging (Bootstrap Aggregating) - the variance-reduction proof

Draw $B$ bootstrap samples (sampling $n$ points **with replacement** from
the original $n$-point training set), train one model $\hat f_b$ per
sample, average: $\hat f_{bag}(x)=\frac1B\sum_b \hat f_b(x)$.

**Derivation of why averaging reduces variance:** if each $\hat f_b(x)$ has
variance $\sigma^2$ and pairwise correlation $\rho$, then:
$$\text{Var}\left(\frac1B\sum_b \hat f_b(x)\right) = \frac{1}{B^2}\left[B\sigma^2 + B(B-1)\rho\sigma^2\right] = \rho\sigma^2 + \frac{(1-\rho)\sigma^2}{B}$$

As $B\to\infty$, variance approaches $\rho\sigma^2$ - **not zero** - the
floor is set entirely by how correlated the trees are. This is precisely
why Random Forest adds **random feature subsetting** at each split (on
top of bagging): it explicitly decorrelates the trees (lowers $\rho$),
pushing the variance floor down further than bagging alone.

### Worked numerical example

Suppose individual tree variance $\sigma^2=4.0$, correlation $\rho=0.3$,
$B=10$ trees:
$$\text{Var}(\hat f_{bag}) = 0.3(4.0) + \frac{0.7(4.0)}{10} = 1.2 + 0.28 = 1.48$$
vs. a single tree's variance of 4.0 - bagging cut variance by **63%**, but
the $\rho\sigma^2=1.2$ floor means more trees beyond this point help very
little (adding $B=100$ instead: $1.2+0.028=1.228$, barely lower).
