## 4. Stacking — brief formalization

Train $L$ base learners $h_1,\dots,h_L$ on the training set (often via
k-fold, to avoid leakage), then train a **meta-learner** $g$ on the
base learners' out-of-fold predictions: $\hat y = g(h_1(x),\dots,h_L(x))$.
Key practical point (deserves its own callout box in the notebook): the
meta-learner MUST be trained on out-of-fold predictions, not
in-sample predictions, or it will overfit to each base learner's training
error rather than learning how to combine genuine generalization behavior.
