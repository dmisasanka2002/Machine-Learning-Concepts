##  Lasso Regression ($\ell_1$ penalty) - why it's different (no closed form)

$$J(\mathbf w) = \|\mathbf y-X\mathbf w\|_2^2 + \lambda\|\mathbf w\|_1,\qquad \|\mathbf w\|_1=\sum_j|w_j|$$

$|w_j|$ is **not differentiable at 0**, so there's no closed-form solution
like Ridge - solved via coordinate descent or subgradient methods.
**Soft-thresholding derivation** (single-feature case, standardized
$X$, i.e. $X^TX=I$): the per-coordinate update minimizes
$(\hat w_j^{OLS}-w_j)^2 + \lambda|w_j|$, whose solution is the
soft-thresholding operator:
$$w_j^{Lasso} = \text{sign}(\hat w_j^{OLS})\cdot\max(|\hat w_j^{OLS}|-\lambda/2,\ 0)$$

### Worked numerical example

Suppose (standardized-feature) OLS coefficients are $\hat w^{OLS}=(0.8,\ -0.3,\ 0.05)$
and $\lambda=0.4$ (so threshold $=\lambda/2=0.2$):
- $w_1 = \text{sign}(0.8)\max(0.8-0.2,0)=0.6$
- $w_2 = \text{sign}(-0.3)\max(0.3-0.2,0)=-0.1$
- $w_3 = \text{sign}(0.05)\max(0.05-0.2,0)=0$ ← **shrunk exactly to zero**

This is the key qualitative difference from Ridge: Lasso performs
**automatic feature selection** because the $\ell_1$ penalty's corners (at
$w_j=0$) make zero an attracting point; $\ell_2$'s smooth bowl never does.
