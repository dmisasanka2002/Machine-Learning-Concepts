## Elastic Net - combining both

$$J(\mathbf w)=\|\mathbf y-X\mathbf w\|_2^2 + \lambda_1\|\mathbf w\|_1 + \lambda_2\|\mathbf w\|_2^2,\quad \text{often reparam. } \alpha\lambda\|\mathbf w\|_1+\tfrac{(1-\alpha)}{2}\lambda\|\mathbf w\|_2^2$$
Handles the case Lasso struggles with: groups of correlated features
(Lasso tends to arbitrarily pick one and zero the rest; the $\ell_2$
component encourages correlated features to be selected/shrunk together).

## 5. Python - verify the hand-worked Ridge example, plus Lasso path

```python
import numpy as np
from sklearn.linear_model import Ridge, Lasso, ElasticNet

X = np.array([[1,1],[1,2],[1,3]])
y = np.array([1,2,2])

ridge = Ridge(alpha=1.0, fit_intercept=False).fit(X, y)
print(ridge.coef_)   # matches hand-derived [0.375, 0.583]

# Lasso soft-thresholding illustration on standardized synthetic data
from sklearn.preprocessing import StandardScaler
rng = np.random.default_rng(0)
Xs = rng.normal(size=(200, 3))
Xs[:,1] = Xs[:,0]*0.05 + rng.normal(scale=0.3, size=200)  # weak signal
y2 = 3*Xs[:,0] - 1*Xs[:,1] + 0.05*Xs[:,2] + rng.normal(scale=0.5, size=200)
for lam in [0.0, 0.1, 0.4, 1.0]:
    m = Lasso(alpha=lam/2).fit(Xs, y2)   # sklearn's alpha = lambda/2 convention differs; check docs
    print(lam, m.coef_)
```
