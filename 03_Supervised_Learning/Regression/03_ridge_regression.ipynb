## 1. Why regularization at all

Ordinary least squares can overfit when $d$ is large relative to $n$, or
when features are collinear ($X^TX$ becomes near-singular, so
$(X^TX)^{-1}$ blows up — direct link to the Linear Algebra primer's
invertibility discussion). Regularization adds a penalty on weight
magnitude to the loss, trading a little bias for a lot less variance
(direct link to `07_ML_Theory/04_bias_variance_tradeoff.ipynb`).

## 2. Ridge Regression ($\ell_2$ penalty) — full derivation

$$J(\mathbf w) = \|\mathbf y - X\mathbf w\|_2^2 + \lambda\|\mathbf w\|_2^2$$

Expand and differentiate (using the matrix-calculus identities from the
Linear Algebra primer §5):
$$J(\mathbf w) = \mathbf y^T\mathbf y - 2\mathbf w^TX^T\mathbf y + \mathbf w^TX^TX\mathbf w + \lambda\mathbf w^T\mathbf w$$
$$\nabla_{\mathbf w} J = -2X^T\mathbf y + 2X^TX\mathbf w + 2\lambda\mathbf w = 0$$
$$\Rightarrow (X^TX + \lambda I)\mathbf w = X^T\mathbf y \Rightarrow \boxed{\hat{\mathbf w}_{ridge} = (X^TX+\lambda I)^{-1}X^T\mathbf y}$$

Note $X^TX+\lambda I$ is **always invertible** for $\lambda>0$ even if
$X^TX$ alone is singular (adding $\lambda I$ shifts every eigenvalue up by
$\lambda$, so none can be zero) — this is the precise mathematical reason
Ridge fixes the collinearity problem, not just an empirical claim.

### Worked numerical example

$X=\begin{bmatrix}1&1\\1&2\\1&3\end{bmatrix}$ (intercept column + one
feature), $\mathbf y=(1,2,2)^T$, $\lambda=1$ (regularize slope only, in
practice — here we regularize both for simplicity of hand computation).

$$X^TX=\begin{bmatrix}3&6\\6&14\end{bmatrix},\quad X^TX+I=\begin{bmatrix}4&6\\6&15\end{bmatrix}$$
$\det=4(15)-6(6)=60-36=24$
$$(X^TX+I)^{-1}=\frac{1}{24}\begin{bmatrix}15&-6\\-6&4\end{bmatrix}$$
$$X^T\mathbf y = \begin{bmatrix}1+2+2\\1+4+6\end{bmatrix}=\begin{bmatrix}5\\11\end{bmatrix}$$
$$\hat{\mathbf w}_{ridge}=\frac{1}{24}\begin{bmatrix}15(5)+(-6)(11)\\-6(5)+4(11)\end{bmatrix}=\frac{1}{24}\begin{bmatrix}75-66\\-30+44\end{bmatrix}=\frac{1}{24}\begin{bmatrix}9\\14\end{bmatrix}=\begin{bmatrix}0.375\\0.583\end{bmatrix}$$

Compare to unregularized OLS on the same data (solve $X^TX\mathbf w=X^T y$):
$\det(X^TX)=3(14)-6(6)=42-36=6$,
$(X^TX)^{-1}=\frac16\begin{bmatrix}14&-6\\-6&3\end{bmatrix}$,
$\mathbf w_{OLS}=\frac16\begin{bmatrix}14(5)-6(11)\\-6(5)+3(11)\end{bmatrix}=\frac16\begin{bmatrix}70-66\\-30+33\end{bmatrix}=\frac16\begin{bmatrix}4\\3\end{bmatrix}=\begin{bmatrix}0.667\\0.5\end{bmatrix}$.
Ridge's slope (0.583) is pulled toward zero less predictably than the
intercept — illustrating that $\ell_2$ shrinks weights but never to
exactly zero.
