# **Generalized Linear Models**

**GLM structure**: $g(E[y\mid x]) = w^Tx$, for **link function** $g$.
- Linear Regression: $g(\mu)=\mu$ (identity link), Gaussian response
- Logistic Regression: $g(\mu)=\log\frac{\mu}{1-\mu}$ (logit link), Bernoulli response
- **Poisson Regression** (new, for count data): $g(\mu)=\log\mu$ (log
  link), Poisson response — used for count outcomes (number of claims,
  number of events), never negative, right-skewed — a natural extension
  worth introducing explicitly given `datasets/insurance.csv` in this
  repo has exactly this kind of count-adjacent structure in some real
  insurance modeling contexts.
This is the explicit unifying insight: Linear and Logistic Regression are
not two unrelated algorithms — they're the **same GLM framework** with
different link functions and response distributions (chosen via MLE on
the appropriate exponential-family distribution, same machinery as
`00_Mathematics\04_statistics.ipynb`).
