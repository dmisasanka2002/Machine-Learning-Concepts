# **Linear Regression**

Linear regression is a fundamental algorithm in machine learning and statistics used for modeling the relationship between a dependent variable and one or more independent variables. It assumes a linear relationship between the variables.

---

## Key Concepts


### **Types of Linear Regression**

- **Simple Linear Regression**:
  - Models the relationship between one independent variable and a dependent variable.
  - Formula: \( y = mx + b \), where:
    - \( m \): Slope (coefficient)
    - \( b \): Intercept

- **Multiple Linear Regression**:
  - Models the relationship between multiple independent variables and a dependent variable.
  - Formula: \( y = b_0 + b_1x_1 + b_2x_2 + ... + b_nx_n \), where:
    - \( b_0 \): Intercept
    - \( b_1, b_2, ..., b_n \): Coefficients of independent variables.

---

## Assumptions of Linear Regression


1. **Linearity**: The relationship between the independent and dependent variable is linear.
2. **Independence**: Observations are independent of each other.
3. **Homoscedasticity**: Constant variance of errors.
4. **Normality**: Errors are normally distributed.
5. **No Multicollinearity**: Independent variables are not highly correlated.

---

## Mathematical Formulation


The goal of linear regression is to minimize the **sum of squared residuals (errors)** between the predicted values and the actual values. This is achieved by solving:

\[
$J(\mathbf{w}) = \frac{1}{2m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})^2$
\]

Where:
- \( $J(\mathbf{w})$ \): Cost function
- \( $\hat{y}^{(i)}$ \): Predicted value
- \( $y^{(i)}$ \): Actual value
- \( $m$\): Number of data points

## Python Implementation


### Simple Linear Regression


```python
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt

# Sample Data
X = np.array([[1], [2], [3], [4], [5]])  # Independent variable
y = np.array([2, 4, 6, 8, 10])          # Dependent variable

# Model Training
model = LinearRegression()
model.fit(X, y)

# Predictions
predictions = model.predict(X)
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)

# Plot Results
plt.scatter(X, y, color='blue', label='Actual')
plt.plot(X, predictions, color='red', label='Predicted')
plt.legend()
plt.show()
```

# **FULL DEEP VERSION with Mathematics**

This supersedes the lighter Linear Regression additions implied earlier and
matches the exact depth bar from your reference doc's own worked example
for this algorithm. Every derivable algorithm in the repo should hit this
bar where mathematically applicable (not every algorithm has an SVD
solution or Gauss-Markov assumptions, but the *spirit* - geometric view +
closed form + numerical stability + statistical assumptions + regularization

**Prerequisites**: Linear Algebra (esp. eigenvalues, projections below),
Calculus (Jacobians - not needed here since this is scalar-output, but
Hessians are, for convexity), MLE/MAP Estimation.

## 1. Problem Setup
$$\hat y = X\beta,\quad X\in\mathbb R^{n\times d},\ \beta\in\mathbb R^d,\ y\in\mathbb R^n$$
$$J(\beta) = \tfrac12\|X\beta-y\|_2^2$$

## 2. Derivation of the Normal Equation

$$J(\beta) = \tfrac12(X\beta-y)^T(X\beta-y) = \tfrac12\big[\beta^TX^TX\beta - 2\beta^TX^Ty + y^Ty\big]$$
$$\nabla_\beta J = X^TX\beta - X^Ty = X^T(X\beta-y)$$
Set to zero: $X^TX\beta = X^Ty$ (the **normal equations**). If $X^TX$ is
invertible: $\boxed{\hat\beta=(X^TX)^{-1}X^Ty}$.

**Second-order check (convexity):** the Hessian is $\nabla^2_\beta J = X^TX$,
which is positive semi-definite for any $X$ (since $v^TX^TXv=\|Xv\|^2\ge0$
for all $v$) - so $J$ is convex and any stationary point is a **global**
minimum, not just local. This is why Linear Regression has no local-minima
problem, unlike neural networks.

## 3. Geometric Interpretation - Projection onto the Column Space

$X\hat\beta$ is the point in $\text{col}(X)$ (the subspace spanned by $X$'s
columns) closest to $y$ in Euclidean distance. The residual
$e=y-X\hat\beta$ is **orthogonal** to $\text{col}(X)$:
$$X^Te = X^T(y-X\hat\beta) = X^Ty - X^TX(X^TX)^{-1}X^Ty = X^Ty-X^Ty=0$$
This orthogonality is *why* the normal equation looks the way it does -
solving $X^TX\beta=X^Ty$ is literally solving "make the residual
perpendicular to every column of $X$." The projection matrix (hat matrix)
is $H=X(X^TX)^{-1}X^T$, and $\hat y = Hy$; $H$ is symmetric and idempotent
($H^2=H$) - the algebraic signature of an orthogonal projection.

## 4. Rank Deficiency, Multicollinearity, and the Pseudoinverse

If columns of $X$ are linearly dependent (perfect multicollinearity) or
$n<d$, $X^TX$ is singular - $(X^TX)^{-1}$ doesn't exist, and the normal
equation has **infinitely many** solutions. The general solution uses the
**Moore-Penrose pseudoinverse** $X^+$, computed via the SVD.

### SVD-based solution - derivation

Write $X = U\Sigma V^T$ ($U\in\mathbb R^{n\times n}$, $V\in\mathbb R^{d\times d}$
orthogonal, $\Sigma$ diagonal with singular values $\sigma_1\ge\dots\ge\sigma_r>0$,
$r=\text{rank}(X)$). Then:
$$X^TX = V\Sigma^TU^TU\Sigma V^T = V\Sigma^T\Sigma V^T$$
The **minimum-norm** least-squares solution is:
$$\hat\beta = X^+y = V\Sigma^+U^Ty,\qquad \Sigma^+_{ii}=1/\sigma_i \text{ if } \sigma_i>0,\text{ else } 0$$
This is exactly what `numpy.linalg.lstsq` / `numpy.linalg.pinv` use
internally, and it always exists - even when $X^TX$ is singular - because
zero singular values are simply mapped to zero (not inverted, avoiding
division by zero). This is also the numerically stable way to solve the
problem - explicitly computing $(X^TX)^{-1}$ squares the condition number
of $X$ ($\kappa(X^TX)=\kappa(X)^2$), amplifying floating-point error; SVD
avoids ever forming $X^TX$ explicitly.

**Multicollinearity in practice:** near-(not exactly-)dependent columns
give $X^TX$ a very small but nonzero smallest eigenvalue, so
$(X^TX)^{-1}$ is technically defined but numerically huge and unstable -
tiny changes in data cause wild swings in $\hat\beta$. This is the precise
mechanism (not just a rule of thumb) behind why Ridge Regression (`03_Supervised_Learning\Regression\04_ridge_regression.ipynb`) fixes it: adding $\lambda I$ raises every
eigenvalue by $\lambda$, so the smallest one is no longer near zero.

## 5. Gauss-Markov Assumptions and the Statistical Interpretation

OLS is BLUE (**B**est **L**inear **U**nbiased **E**stimator) under:
1. Linearity: $y=X\beta+\epsilon$
2. Strict exogeneity: $E[\epsilon\mid X]=0$
3. No perfect multicollinearity: $X$ has full column rank
4. Homoscedasticity: $\text{Var}(\epsilon_i\mid X)=\sigma^2$ (constant)
5. No autocorrelation: $\text{Cov}(\epsilon_i,\epsilon_j)=0,\ i\neq j$

Under these, $E[\hat\beta]=\beta$ (unbiased - plug $y=X\beta+\epsilon$ into
$\hat\beta=(X^TX)^{-1}X^Ty$ and take expectation: the $\epsilon$ term
vanishes) and $\text{Var}(\hat\beta)=\sigma^2(X^TX)^{-1}$, the **minimum**
variance among all linear unbiased estimators (Gauss-Markov theorem).
Violating homoscedasticity/no-autocorrelation doesn't bias $\hat\beta$ but
makes it no longer minimum-variance, and makes the standard errors used
for hypothesis tests on $\beta$ wrong - this is why residual diagnostic
plots (§7 below) matter in practice, not just as a checkbox.

## 6. Worked Numerical Example (small enough for full hand computation)

$$X=\begin{bmatrix}1&1\\1&2\\1&3\end{bmatrix},\quad y=\begin{bmatrix}2\\3\\5\end{bmatrix}$$
$$X^TX=\begin{bmatrix}3&6\\6&14\end{bmatrix},\qquad X^Ty=\begin{bmatrix}2+3+5\\2+6+15\end{bmatrix}=\begin{bmatrix}10\\23\end{bmatrix}$$
$\det(X^TX)=3(14)-6(6)=42-36=6$
$$(X^TX)^{-1}=\frac16\begin{bmatrix}14&-6\\-6&3\end{bmatrix}$$
$$\hat\beta = \frac16\begin{bmatrix}14(10)-6(23)\\-6(10)+3(23)\end{bmatrix}=\frac16\begin{bmatrix}140-138\\-60+69\end{bmatrix}=\frac16\begin{bmatrix}2\\9\end{bmatrix}=\begin{bmatrix}0.333\\1.5\end{bmatrix}$$

Predictions: $\hat y = X\hat\beta = \{0.333+1.5(1),\ 0.333+1.5(2),\ 0.333+1.5(3)\}=\{1.833,\ 3.333,\ 4.833\}$

Residuals: $e=y-\hat y=\{0.167,\ -0.333,\ 0.167\}$ 

- check orthogonality:

$X^Te = \{0.167-0.333+0.167,\ 0.167-0.667+0.5\} = \{0.001\approx0,\ 0.0\}$ ✓ (rounding)

MSE $=\frac13(0.167^2+0.333^2+0.167^2)=\frac13(0.0279+0.1109+0.0279)=0.0556$

$\bar y = 10/3=3.333$; $SS_{tot}=\sum(y_i-\bar y)^2=(2-3.333)^2+(3-3.333)^2+(5-3.333)^2=1.777+0.111+2.778=4.667$

$SS_{res}=\sum e_i^2=0.0279+0.1109+0.0279=0.1667$

$$R^2 = 1-\frac{SS_{res}}{SS_{tot}} = 1-\frac{0.1667}{4.667}=1-0.0357=0.964$$

````python
## verify every hand-computed number above

import numpy as np
from sklearn.linear_model import LinearRegression

X = np.array([[1,1],[1,2],[1,3]])
y = np.array([2,3,5])

XtX = X.T @ X
Xty = X.T @ y
beta_hat = np.linalg.inv(XtX) @ Xty
print("beta:", beta_hat)                      # [0.333, 1.5]

y_hat = X @ beta_hat
resid = y - y_hat
print("residuals:", resid)                    # [0.167, -0.333, 0.167]
print("X^T e (should be ~0):", X.T @ resid)    # orthogonality check

mse = np.mean(resid**2)
r2 = 1 - (resid**2).sum() / ((y - y.mean())**2).sum()
print("MSE:", mse, "R2:", r2)                  # 0.0556  0.964

# cross-check against sklearn and the SVD/pseudoinverse solution
sk = LinearRegression(fit_intercept=False).fit(X, y)
print("sklearn beta:", sk.coef_)
print("pinv beta:", np.linalg.pinv(X) @ y)     # identical when X has full column rank
````

## 7. Complexity Analysis

| Method | Time | Space |
|---|---|---|
| Normal equation | $O(nd^2+d^3)$ | $O(d^2)$ |
| SVD / pseudoinverse | $O(nd^2)$ (typically, for $n\ge d$) | $O(nd)$ |
| Gradient Descent | $O(nd)$ per iteration | $O(d)$ |
Normal equation is preferred for small-to-moderate $d$ (say, $d<10{,}000$);
gradient descent is preferred when $d$ is very large, since forming/inverting
a $d\times d$ matrix becomes prohibitive.
