### 1. The Kernel Trick - motivation and derivation

Many algorithms (SVM, PCA, ridge regression) can be written using only

**inner products** between data points, $x_i\cdot x_j$, never the raw
features themselves (true for SVM's dual form — see `13_svm.ipynb`).
This means: to make the algorithm nonlinear, instead of explicitly mapping
$x\to\phi(x)$ into a higher-dimensional space and computing
$\phi(x_i)\cdot\phi(x_j)$ (potentially very expensive, even infinite-dimensional),
just replace the inner product with a **kernel function**
$K(x_i,x_j)=\phi(x_i)\cdot\phi(x_j)$ computed **directly in the original
space** — often $O(d)$ instead of $O(\text{dim}(\phi))$.

**Mercer's theorem** (stated, not proved — genuinely graduate-level
functional analysis): any function $K(x,y)$ that is symmetric and produces
a positive-semi-definite Gram matrix for any finite sample is a valid
kernel — i.e. corresponds to *some* inner product in *some* (possibly
infinite-dimensional) feature space, even if $\phi$ is never written down
explicitly.

### 2. Worked example - the polynomial kernel, explicit $\phi$

For $x=(x_1,x_2)\in\mathbb R^2$, degree-2 kernel $K(x,z)=(x\cdot z)^2$.
Expand: $(x\cdot z)^2 = (x_1z_1+x_2z_2)^2 = x_1^2z_1^2+2x_1x_2z_1z_2+x_2^2z_2^2$.
This equals $\phi(x)\cdot\phi(z)$ for $\phi(x)=(x_1^2,\sqrt2\,x_1x_2,x_2^2)$
— an **explicit** 3-dimensional feature map, verifiable by direct
substitution: $\phi(x)\cdot\phi(z) = x_1^2z_1^2+2x_1x_2z_1z_2+x_2^2z_2^2$ ✓
matches exactly. This concretely demonstrates the trick: compute
$(x\cdot z)^2$ in $O(2)$ operations in the original 2D space instead of
$O(3)$ after an explicit map — the saving grows dramatically for
higher-degree polynomials or the RBF kernel (which corresponds to an
**infinite**-dimensional $\phi$, so explicit mapping is literally
impossible, and the kernel trick is the *only* way to use it).

### 3. RBF (Gaussian) Kernel

$$K(x,z) = \exp\left(-\frac{\|x-z\|^2}{2\sigma^2}\right)$$
Corresponds to an infinite-dimensional feature space (expand the Taylor
series of the exponential to see infinitely many polynomial terms of all
degrees) — this is why RBF-kernel SVMs can fit extremely flexible
nonlinear boundaries.

```python
## verify the polynomial kernel = explicit feature map numerically

import numpy as np

x = np.array([1.0, 2.0]); z = np.array([3.0, 1.0])
kernel_direct = (x @ z)**2

def phi(v): return np.array([v[0]**2, np.sqrt(2)*v[0]*v[1], v[1]**2])
kernel_via_phi = phi(x) @ phi(z)

print(kernel_direct, kernel_via_phi)   # both 25.0 -- identical
```
