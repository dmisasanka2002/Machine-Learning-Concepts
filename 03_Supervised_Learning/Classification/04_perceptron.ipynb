# **Perceptron**

### 1. Algorithm and Update Rule - derivation

Linear classifier $\hat y = \text{sign}(w^Tx+b)$. On seeing a **misclassified**
point $(x_i,y_i)$ (i.e. $y_i(w^Tx_i+b)\le0$), update:
$$w \leftarrow w + \eta\, y_i x_i, \qquad b\leftarrow b+\eta\, y_i$$
**Why this update makes sense**: after the update, the new margin on that
point is $y_i((w+\eta y_ix_i)^Tx_i+b) = y_i(w^Tx_i+b) + \eta y_i^2\|x_i\|^2 = y_i(w^Tx_i+b)+\eta\|x_i\|^2$
— strictly larger than before the update (since $\eta\|x_i\|^2>0$), moving
the point toward being correctly classified. This is literally gradient
descent on the **hinge-at-zero loss** $\max(0,-y_i(w^Tx_i+b))$.

### 2. Perceptron Convergence Theorem (statement)

If the data is linearly separable with margin $\gamma$ (there exists a
unit-norm $w^*$ such that $y_i(w^{*T}x_i)\ge\gamma$ for all $i$), the
Perceptron algorithm makes at most $(R/\gamma)^2$ mistakes before
converging, where $R=\max_i\|x_i\|$. **Crucially**: it does NOT converge
if the data is not linearly separable (it will cycle forever) — direct
motivation for why SVM (which finds the *maximum*-margin separator and
handles non-separable data via slack variables) superseded the Perceptron
historically, and why multi-layer (nonlinear) networks were needed for
XOR-like problems (ties directly to the VC-dimension/XOR
non-separability result already proved in `07_ML_Theory\04_generalization_and_capacity.ipynb`).

### 3. Worked numerical example - trace 2 updates by hand

Data: $x_1=(1,1), y_1=+1$; $x_2=(-1,-1), y_2=-1$; $x_3=(1,-1),y_3=+1$
(linearly separable). Init $w=(0,0)$, $b=0$, $\eta=1$.
- Check $x_1$: $y_1(w\cdot x_1+b)=1(0)=0\le0$ → misclassified → update:
  $w=(0,0)+1(1)(1,1)=(1,1)$, $b=0+1(1)=1$
- Check $x_2$: $y_2(w\cdot x_2+b)=-1[(1)(-1)+(1)(-1)+1]=-1[-1]=1>0$ → correct, no update
- Check $x_3$: $y_3(w\cdot x_3+b)=1[(1)(1)+(1)(-1)+1]=1[1]=1>0$ → correct, no update
Converged after just 1 update: $w=(1,1)$, $b=1$.

```python
## verify the hand-worked Perceptron trace

import numpy as np
X = np.array([[1,1],[-1,-1],[1,-1]])
y = np.array([1,-1,1])
w, b, eta = np.zeros(2), 0.0, 1.0
for i in range(len(X)):
    if y[i]*(w@X[i]+b) <= 0:
        w += eta*y[i]*X[i]
        b += eta*y[i]
print(w, b)   # [1. 1.] 1.0 -- matches hand trace
```
