# **Quadratic Discriminant Analysis**

## 1. QDA - why the boundary becomes quadratic

Drop the shared-covariance assumption: each class has its own $\Sigma_k$.
Now the $x^T\Sigma_k^{-1}x$ term does **not** cancel between classes (since
$\Sigma_k$ differs), leaving a term quadratic in $x$ in the discriminant
function:
$$\delta_k(x) = -\tfrac12\log|\Sigma_k| - \tfrac12(x-\mu_k)^T\Sigma_k^{-1}(x-\mu_k) + \log P(y=k)$$
Decision boundaries $\delta_j(x)=\delta_k(x)$ are now **quadratic**
surfaces (ellipses/parabolas/hyperbolas in 2D) — QDA can fit more flexible
boundaries than LDA at the cost of estimating $K$ separate $d\times d$
covariance matrices instead of one shared one (more parameters → more
variance, a direct bias-variance tradeoff between LDA and QDA worth
stating explicitly).

## 2. Complexity / Parameter Count (ties to Bias-Variance directly)

LDA estimates $K$ means + 1 shared $\Sigma$ ($O(Kd + d^2)$ parameters).
QDA estimates $K$ means + $K$ separate $\Sigma_k$ ($O(Kd + Kd^2)$
parameters) — grows much faster with $d$, making QDA more prone to
overfitting in high dimensions/small $n$ — precisely the bias-variance
tradeoff from `04-Bias-Variance-Tradeoff.md` applied to a concrete model
choice.

## 3. Python - verify the hand-worked LDA boundary and compare LDA vs QDA regions

```python
import numpy as np
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis

X = np.array([[1],[2],[3],[5],[6],[7]])
y = np.array([0,0,0,1,1,1])
lda = LinearDiscriminantAnalysis().fit(X, y)
xs = np.linspace(0, 8, 200).reshape(-1,1)
preds = lda.predict(xs)
boundary = xs[np.where(np.diff(preds)!=0)[0][0]][0]
print("LDA boundary approx:", boundary)   # ~4.0, matches hand derivation
```
