# **Linear Discriminant Analysis**

Prerequisites: Probability primer (multivariate Gaussian), Naive Bayes
(same Bayes'-rule generative framework, different independence assumption).

## 1. Shared Setup - Generative Classification via Bayes' Rule

Both LDA and QDA model $P(x\mid y=k)$ as multivariate Gaussian and classify
via $\arg\max_k P(y=k\mid x) = \arg\max_k P(x\mid y=k)P(y=k)$ (Bayes' rule,
denominator dropped since it's the same for all $k$) — same framework as
Naive Bayes, but **without** the feature-independence assumption; instead
using the full covariance structure.

## 2. LDA - derivation of the linear decision boundary

Assume **shared** covariance $\Sigma$ across all classes:
$$P(x\mid y=k) = \frac{1}{(2\pi)^{d/2}|\Sigma|^{1/2}}\exp\left(-\tfrac12(x-\mu_k)^T\Sigma^{-1}(x-\mu_k)\right)$$
Taking logs of $P(x|y=k)P(y=k)$ and dropping terms constant across $k$
(the $(2\pi)^{d/2}|\Sigma|^{1/2}$ normalizer, and the $x^T\Sigma^{-1}x$
term — which is now identical for every class since $\Sigma$ is shared):
$$\delta_k(x) = x^T\Sigma^{-1}\mu_k - \tfrac12\mu_k^T\Sigma^{-1}\mu_k + \log P(y=k)$$
This **discriminant function** $\delta_k(x)$ is **linear in $x$** — hence
"Linear" Discriminant Analysis. The decision boundary between classes $j,k$
is where $\delta_j(x)=\delta_k(x)$, which is a hyperplane (linear).

### Worked numerical example (2 classes, 1 feature, for hand-tractability)

Class 0: $\mu_0=2$, Class 1: $\mu_1=6$, shared $\sigma^2=4$, equal priors
$P(y=0)=P(y=1)=0.5$. Boundary where $\delta_0(x)=\delta_1(x)$:
$$\frac{x\mu_0}{\sigma^2}-\frac{\mu_0^2}{2\sigma^2} = \frac{x\mu_1}{\sigma^2}-\frac{\mu_1^2}{2\sigma^2}$$
$$\frac{x(2)}{4}-\frac{4}{8} = \frac{x(6)}{4}-\frac{36}{8} \Rightarrow 0.5x-0.5 = 1.5x-4.5 \Rightarrow 4=x \Rightarrow x^*=4$$
The boundary sits exactly halfway between the means (2 and 6) since priors
and variances are equal — a sanity-checkable result: classify $x<4$ as
class 0, $x>4$ as class 1.
