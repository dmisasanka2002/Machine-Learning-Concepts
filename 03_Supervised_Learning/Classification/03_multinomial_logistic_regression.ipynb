# **Multinomial Logistic Regression**

## Softmax derivation

$$P(y=k\mid x) = \frac{e^{w_k^Tx}}{\sum_{j=1}^K e^{w_j^Tx}}$$
Cross-entropy loss generalizes directly:
$$J(W) = -\sum_i\sum_k \mathbb 1[y_i=k]\log P(y_i=k\mid x_i)$$
Gradient (derived analogously to binary Logistic Regression's
$X^T(\sigma(Xw)-y)$ in `03_Supervised_Learning\Classification\01_logistic_regression.ipynb` *Gradient Derivation*):
$$\nabla_{w_k}J = \sum_i (P(y_i=k\mid x_i) - \mathbb1[y_i=k])\,x_i$$
— same "predicted minus actual, times input" pattern appearing a third
time in this repo (Linear Regression, binary Logistic Regression, now
multinomial) — worth calling out explicitly as a recurring structural
insight rather than three unrelated formulas to memorize.
