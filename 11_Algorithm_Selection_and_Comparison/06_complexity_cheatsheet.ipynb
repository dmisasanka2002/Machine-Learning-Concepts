# Master Complexity Cheatsheet

Notation: $n$ = #samples, $d$ = #features, $k$ = #neighbors/clusters,
$T$ = #trees/boosting rounds, $D$ = tree depth, $s$ = #support vectors.

| Algorithm | Training Time | Prediction Time (per sample) | Space | Derivation note |
|---|---|---|---|---|
| Linear Regression (Normal Eq.) | $O(nd^2 + d^3)$ | $O(d)$ | $O(d^2)$ | $X^TX$ costs $O(nd^2)$; inverting a $d\times d$ matrix costs $O(d^3)$ |
| Linear Regression (Gradient Descent) | $O(n d \cdot I)$, $I$=iterations | $O(d)$ | $O(d)$ | each iteration is one pass over $X$: $O(nd)$ |
| Logistic Regression (GD) | $O(nd \cdot I)$ | $O(d)$ | $O(d)$ | identical structure to linear regression GD |
| KNN | $O(1)$ (lazy) | $O(nd)$ naive, $O(d\log n)$ with KD-tree (low $d$) | $O(nd)$ | must compute distance to all $n$ points at query time |
| Naive Bayes | $O(nd)$ | $O(cd)$, $c$=#classes | $O(cd)$ | one pass to count/estimate per-feature statistics |
| Decision Tree | $O(n d \log n)$ | $O(D)$, $D\approx\log n$ | $O(n)$ (nodes) | at each of $\log n$ depth levels, sorting/scanning costs $O(nd)$ |
| Random Forest | $O(T\cdot n d \log n)$ | $O(TD)$ | $O(Tn)$ | $T$ independent trees, trivially parallelizable |
| AdaBoost | $O(T\cdot n d)$ (stumps) | $O(T)$ | $O(T)$ | $T$ sequential weak learners, each $O(nd)$ to fit a stump |
| Gradient Boosting | $O(T\cdot n d \log n)$ | $O(TD)$ | $O(Tn)$ | like RF but sequential (can't parallelize across trees) |
| SVM (SMO/QP solver) | $O(n^2)$ to $O(n^3)$ | $O(s\cdot d)$ | $O(n^2)$ (kernel matrix) or $O(sd)$ | quadratic program over $n$ dual variables |
| K-Means | $O(nkd \cdot I)$ | $O(kd)$ | $O(nd + kd)$ | each iteration: assign ($O(nkd)$) + update ($O(nd)$) |
| Hierarchical (naive) | $O(n^3)$, $O(n^2\log n)$ optimized | — (not inductive) | $O(n^2)$ | must maintain/update full $n\times n$ distance matrix |
| DBSCAN | $O(n\log n)$ with spatial index, $O(n^2)$ naive | — | $O(n)$ | region query per point; index brings it down from $n^2$ |
| PCA | $O(nd^2 + d^3)$ (via covariance eigen-decomp), $O(nd\cdot k)$ via SVD for top $k$ | $O(dk)$ | $O(d^2)$ or $O(dk)$ | eigen-decomposition of the $d\times d$ covariance matrix |
| GMM (EM) | $O(nkd^2 \cdot I)$ | $O(kd^2)$ | $O(kd^2)$ | each E-step evaluates $n\times k$ multivariate Gaussians, each $O(d^2)$ (or $O(d^3)$ with full covariance inversion) |
| t-SNE | $O(n^2)$ naive, $O(n\log n)$ Barnes-Hut | — (not inductive) | $O(n^2)$ or $O(n)$ | pairwise affinities between all points |

**How to read/derive one row (worked example — Decision Tree):**

At each level of the tree (there are $O(\log n)$ levels for a roughly
balanced tree), the algorithm must consider every feature ($d$) and, for
each, sort or scan the $n$ samples at that level to find the best split
threshold: $O(nd)$ work per level $\Rightarrow$ total training cost
$O(nd\log n)$. Prediction just walks root→leaf: $O(D)\approx O(\log n)$.
This exact reasoning pattern (cost-per-level × number-of-levels, or
cost-per-iteration × number-of-iterations) is what every notebook's §5
Complexity Analysis should show — not just state the final Big-O.
