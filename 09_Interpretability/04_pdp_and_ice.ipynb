# **PDP (Partial Dependence Plot)**

- average predicted output as one feature $x_j$
varies, marginalizing over all other features:
$$\text{PDP}_j(v) = \frac1n\sum_{i=1}^n \hat f(x_j=v,\ x_{i,-j})$$
(replace feature $j$ with value $v$ for every training point, average the
predictions). 

# **ICE (Individual Conditional Expectation)**

- the same computation but plotted **per individual point** rather than averaged - reveals heterogeneous effects (e.g. a feature that helps some people and hurts others) that a PDP's averaging would hide entirely - worth demonstrating with a synthetic interaction-effect dataset where the PDP looks flat but ICE lines fan out in opposite directions.
