# **Validation Strategies**

- **Time-series split**: never use future data to predict the past — folds
  must respect temporal order (`sklearn.model_selection.TimeSeriesSplit`).
- **Group k-fold**: when samples are grouped (e.g. multiple rows per
  patient), ensure a group never spans both train and test — otherwise
  leakage inflates apparent performance (direct link to
  `02_Data/07_data_leakage.ipynb`).
- **Nested CV**: outer loop estimates generalization error, inner loop
  does hyperparameter search — prevents the optimistic bias of using the
  same CV loop for both tuning and evaluation.
