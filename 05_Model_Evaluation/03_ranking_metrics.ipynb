# **Ranking Metrics**

- **Precision@k**: fraction of the top-$k$ ranked items that are relevant.
- **MAP (Mean Average Precision)**: mean, across queries, of the average
  precision (area under the precision-recall curve computed over ranked
  retrieval).
- **NDCG**: discounts relevance by rank position via $\log_2$, rewarding
  relevant items appearing earlier — formula:
  $$\text{DCG@k}=\sum_{i=1}^k \frac{\text{rel}_i}{\log_2(i+1)}, \qquad \text{NDCG@k}=\frac{\text{DCG@k}}{\text{IDCG@k}}$$
  (normalize by the DCG of the ideal ranking).
