# **Calibration**

A classifier is **well-calibrated** if, among all points predicted
probability $\approx p$, the true fraction of positives is also
$\approx p$.

- Naive Bayes is a classic example of a model whose *ranking* of examples can be good while its raw probability outputs are poorly calibrated (its independence assumption tends to push predicted probabilities toward 0 or 1 more than warranted).

**Brier score** (proper scoring rule for calibration + accuracy jointly):
$$\text{Brier} = \frac1n\sum_i (\hat p_i - y_i)^2$$
**Platt scaling / isotonic regression** (LIGHT, practical): fit a
secondary 1-D model (logistic or isotonic) mapping raw scores to
calibrated probabilities, using a held-out calibration set.
