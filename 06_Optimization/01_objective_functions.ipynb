## 1. Theory

Every model in this repo is fit by minimizing (or maximizing) some
scalar function of its parameters - the objective function. Naming and
distinguishing the terms precisely here avoids confusion later:
**loss** = cost of a single prediction (e.g. $(y_i-\hat y_i)^2$);
**cost/objective** = the aggregate over the training set (e.g.
$\frac1n\sum_iL(y_i,\hat y_i)$), often with a regularization term added; **risk** = the *expected* loss over the true
data distribution - unobservable, only estimable via the empirical
cost (formalized precisely in `07_ML_Theory/02_loss_and_risk.ipynb`).

## 2. Objective Functions Already Built in This Repo

| Model | Objective |
|---|---|
| Linear Regression | $\frac12\|X w - y\|_2^2$ (`03_Supervised_Learning\Regression\01_linear_regression.ipynb`) |
| Ridge/Lasso | above + $\lambda$-penalty (`03_Supervised_Learning\Regression\04, 05`) |
| Logistic Regression | cross-entropy (`03_Supervised_Learning\Classification\01_logistic_regression.ipynb`) |
| SVM | $\frac12\|w\|^2$ + hinge-loss slack (`03_Supervised_Learning\Classification\13_svm.ipynb`) |
| K-Means | WCSS (`04_Unsupervised_Learning\01_kmeans.ipynb`) |
| GMM | (negative) log-likelihood, via EM (`04_Unsupervised_Learning\08_gmm_and_em.ipynb`) |


Stating them side by side here is deliberate: despite looking different,
every one of them is "some notion of error/negative-fit, minimized over
parameters" - the unifying idea `07_ML_Theory\01_hypothesis_spaces_and_erm.ipynb`
formalizes as regularized Empirical Risk Minimization.

## 3. Worked Numerical Example

For 3 points $(1,2),(2,3),(3,5)$ and candidate $w=1.5,b=0$: compute the
per-point squared-error loss, then the aggregate cost.

$\hat y = \{1.5, 3.0, 4.5\}$, 

losses $=\{(2-1.5)^2,(3-3.0)^2,(5-4.5)^2\}=\{0.25,0,0.25\}$,


cost $=\frac13(0.5)=0.1\overline6$.

```python
import numpy as np
x = np.array([1,2,3]); y = np.array([2,3,5])
w, b = 1.5, 0
pred = w*x + b
losses = (pred - y)**2
print(losses, losses.mean())   # [0.25, 0., 0.25]  0.1667
```
