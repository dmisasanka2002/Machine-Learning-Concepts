# **Convexity**

**Prerequisites:** Objective Functions, Calculus/Matrix Calculus | **Next:** Gradient Descent | **Depth tier:** Core

## 1. Theory

Whether gradient descent (or any local-search optimizer) is guaranteed to
find the *best possible* answer, or merely *an* answer, depends entirely
on the shape of the objective function - convexity is the precise
property that gives the strong guarantee.

## 2. Mathematical Derivation

**Convex function** definition: $f(\lambda x+(1-\lambda)y)\le\lambda f(x)+(1-\lambda)f(y)$
for all $\lambda\in[0,1]$ - the function lies on or below the straight
line connecting any two of its points. **Second-order characterization**:
$f$ is convex iff its Hessian $\nabla^2f$ is positive semi-definite
everywhere (already used in `00_Mathematics/02_calculus_and_matrix_calculus.ipynb`'s
worked example, which found a *non*-PSD Hessian and correctly concluded
non-convexity at that point).

**Why this matters for ML - proof sketch**: for convex $J(\theta)$, any
local minimum is global. If $\theta^*$ were a local-but-not-global
minimum, the line segment between $\theta^*$ and the true global minimum
would, by the convexity inequality above, produce points with value
strictly between $J(\theta^*)$ and the global minimum's value, arbitrarily
close to $\theta^*$ - contradicting $\theta^*$ being a *local* minimum
(no nearby point should have lower value). Linear Regression's Hessian
$X^TX$ and Logistic Regression's Hessian $X^T\text{diag}(p_i(1-p_i))X$
are both PSD (already shown in `03_Supervised_Learning\Regression\01_linear_regression.ipynb` and
`03_Supervised_Learning\Classification\01_logistic_regression.ipynb`) - this is *why* gradient descent
provably finds the global optimum for those two specific algorithms, not
a general property of every ML model (neural network losses, covered in
the DL repo, are famously non-convex, with many distinct local minima).

## 3. Worked Numerical Example

- Is $f(x)=x^4-3x^2$ convex? 

$f''(x)=12x^2-6$. At $x=0$: 

$f''(0)=-6<0$ → **not convex** there (a local concavity - this function actually has two
separate local minima, one of which is not global, precisely because it's
non-convex 

verify: 

$f'(x)=4x^3-6x=0 \Rightarrow x=0,\pm\sqrt{1.5}$;

$f(\sqrt{1.5})=f(-\sqrt{1.5})=2.25-4.5=-2.25$ at both, so both are
(tied) global minima here, but a slightly asymmetric variant would have
one merely local).

```python
## verify the Hessian sign and find both minima

import numpy as np
from scipy.optimize import minimize

def f(x): return x**4 - 3*x**2
def fpp(x): return 12*x**2 - 6
print(fpp(0))          # -6, confirms non-convex at x=0

res1 = minimize(f, x0=1.0)   # starts near the positive-side minimum
res2 = minimize(f, x0=-1.0)  # starts near the negative-side minimum
print(res1.x, res1.fun, res2.x, res2.fun)  # both land on ~-2.25, tied global minima here
```

## 4. Failure Cases
Applying plain gradient descent to a non-convex objective and reporting
"the" optimum without checking multiple initializations - silently
reports a local, possibly poor, optimum as if it were global.

## 5. Assumptions
The global-optimum guarantee requires both convexity AND a small-enough
learning rate (the Lipschitz-gradient condition in
`06_Optimization/03_gradient_descent.ipynb`) - convexity alone doesn't
guarantee convergence if $\alpha$ is too large.
