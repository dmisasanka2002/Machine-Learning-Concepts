# **Gradient Decent Algorithm**

## Full Derivation

Taylor-expand $J(\theta-\alpha\nabla J(\theta))$ around $\theta$ to first
order: $J(\theta)-\alpha\|\nabla J(\theta)\|^2 + O(\alpha^2)$ - for small
enough $\alpha>0$, this is less than $J(\theta)$, so moving in the
direction $-\nabla J(\theta)$ locally decreases the objective. This is the
one-line justification for "why subtract the gradient," not just a
convention.

**Convergence condition**: for $L$-Lipschitz-gradient convex $J$, GD
converges if $\alpha \le 1/L$ (formal statement, proof omitted at this
course level - cite it, since deriving it fully requires more real
analysis than is proportionate here, consistent with the house "light
treatment" standard for genuinely graduate-level derivations).

### Worked numerical example (already partially shown in the Linear Regression additions - extended here to trace 3 iterations explicitly)

$J(w)=w^2-4w+5$ (a toy convex quadratic, minimum at $w^*=2$ by calculus:
$J'(w)=2w-4=0$). Start $w_0=0$, $\alpha=0.2$.
$$J'(w)=2w-4$$
- $w_1 = 0 - 0.2(2\cdot0-4) = 0-0.2(-4)=0.8$
- $w_2 = 0.8-0.2(2\cdot0.8-4)=0.8-0.2(-2.4)=0.8+0.48=1.28$
- $w_3 = 1.28-0.2(2\cdot1.28-4)=1.28-0.2(-1.44)=1.28+0.288=1.568$

Converging toward $w^*=2$, with the step size shrinking each iteration as
the gradient magnitude shrinks near the optimum - the visual signature of
gradient descent on a convex bowl
