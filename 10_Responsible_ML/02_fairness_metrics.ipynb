## **Fairness Metrics**

- A few actually-defined, formal metrics

- **Demographic parity**: $P(\hat y=1\mid A=0) = P(\hat y=1\mid A=1)$ (equal
  positive-prediction rate across groups $A$).
- **Equalized odds**: $P(\hat y=1\mid y=1,A=a)$ and $P(\hat y=1\mid y=0,A=a)$
  are equal across groups $a$ (equal TPR and FPR across groups).
- **Predictive parity**: $P(y=1\mid \hat y=1, A=a)$ equal across groups
  (equal precision across groups).
**Worth stating explicitly as a real mathematical result**: these three
fairness criteria are, in general, **mutually incompatible** whenever base
rates differ between groups (a known impossibility result, Chouldechova
2017/Kleinberg et al. 2016) — there is no free lunch here either, directly
echoing `07_ML_Theory\07_no_free_lunch.ipynb`'s No-Free-Lunch theme; the choice of
which fairness metric to optimize is a policy decision, not a purely
technical one.
