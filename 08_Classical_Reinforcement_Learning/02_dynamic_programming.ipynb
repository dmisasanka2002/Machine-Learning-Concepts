## Policy Iteration & Value Iteration

**Bellman optimality equation**: $V^*(s)=\max_a\left[R(s,a)+\gamma\sum_{s'}P(s'|s,a)V^*(s')\right]$
**Value iteration**: repeatedly apply this update to all states until
convergence (a contraction mapping — converges by the Banach fixed-point
theorem since $\gamma<1$ makes the Bellman operator a $\gamma$-contraction).

### Worked example: value iteration on the 3-state MDP above, 2 sweeps

Init $V_0=(0,0,0)$ for $(A,B,C)$.
Sweep 1: $V_1(B)=\max[5+0.9(0)]=5$; $V_1(A)=\max[1+0.9(0)]=1$
Sweep 2: $V_2(B)=5+0.9(0)=5$ (unchanged, C always 0); $V_2(A)=1+0.9(5)=5.5$
— converges to the same $V^*$ as the direct policy evaluation above in 2
sweeps for this tiny example (in general converges asymptotically).
