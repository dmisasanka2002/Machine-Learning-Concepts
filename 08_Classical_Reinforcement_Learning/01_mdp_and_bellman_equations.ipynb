**Markov Decision Process**: $(\mathcal S,\mathcal A,P,R,\gamma)$ — states,
actions, transition probabilities $P(s'|s,a)$, reward function $R(s,a)$,
discount factor $\gamma\in[0,1)$.

**Value function**: $V^\pi(s) = E_\pi\left[\sum_{t=0}^\infty \gamma^t R(s_t,a_t)\mid s_0=s\right]$

**Bellman equation — derivation** (recursive decomposition of the sum):
$$V^\pi(s) = E_\pi\Big[R(s,a) + \gamma\sum_{t=1}^\infty \gamma^{t-1}R(s_t,a_t)\Big] = \sum_a \pi(a|s)\left[R(s,a)+\gamma\sum_{s'}P(s'|s,a)V^\pi(s')\right]$$
The second sum is exactly $V^\pi(s_1)$ again, one step later — giving the
self-referential Bellman equation.

### Worked numerical example: tiny 3-state MDP, evaluate a fixed policy

States $\{A,B,C\}$ (C terminal, reward 0 forever after). From A: action
"right" → B with reward 1. From B: action "right" → C with reward 5.
$\gamma=0.9$. Policy $\pi$: always go right.
$$V^\pi(C)=0,\qquad V^\pi(B)=5+0.9(0)=5,\qquad V^\pi(A)=1+0.9(5)=1+4.5=5.5$$
