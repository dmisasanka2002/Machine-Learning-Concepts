## Off-policy TD control, full derivation

$$Q(s_t,a_t)\leftarrow Q(s_t,a_t)+\alpha\Big[R_t+\gamma\max_{a'}Q(s_{t+1},a')-Q(s_t,a_t)\Big]$$
Difference from SARSA: uses $\max_{a'}Q(s_{t+1},a')$ (the best possible next
action) rather than the action actually taken next — makes Q-learning
**off-policy** (learns the optimal policy's values regardless of the
exploration policy actually being followed).

### Worked numerical example — one Q-learning update

$Q(s_1,\text{right})=2.0$ currently. Agent takes action "right", observes
reward $R=3$, lands in $s_2$ where $\max_{a'}Q(s_2,a')=4.0$. $\alpha=0.5$,
$\gamma=0.9$:
$$Q(s_1,\text{right}) \leftarrow 2.0 + 0.5\big[3+0.9(4.0)-2.0\big] = 2.0+0.5[3+3.6-2.0]=2.0+0.5(4.6)=2.0+2.3=4.3$$

## 6. Python — verify the Bellman/value-iteration and Q-learning hand examples

```python
gamma = 0.9
# Direct policy evaluation
V_C, V_B = 0, 5 + gamma*0
V_A = 1 + gamma*V_B
print(V_A, V_B, V_C)   # 5.5 5.0 0

# Value iteration, 2 sweeps
V = {"A":0, "B":0, "C":0}
for sweep in range(2):
    V["B"] = max(5 + gamma*V["C"], V["B"])
    V["A"] = max(1 + gamma*V["B"], V["A"])
print(V)   # {'A': 5.5, 'B': 5.0, 'C': 0}

# Q-learning update
Q_s1_right = 2.0
alpha, R, max_Q_s2 = 0.5, 3, 4.0
Q_s1_right_new = Q_s1_right + alpha*(R + gamma*max_Q_s2 - Q_s1_right)
print(Q_s1_right_new)   # 4.3
```
