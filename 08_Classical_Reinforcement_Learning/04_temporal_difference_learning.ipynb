## TD(0) derivation

Update *after every step* (not waiting for episode end, unlike Monte Carlo)
using the **TD error**:
$$\delta_t = R_t + \gamma V(s_{t+1}) - V(s_t), \qquad V(s_t)\leftarrow V(s_t)+\alpha\,\delta_t$$
This is bootstrapping: using the current (possibly still-wrong) estimate
$V(s_{t+1})$ as a stand-in for the true future return — trades some bias
for much lower variance than Monte Carlo's full-episode return (direct
parallel to the general Bias-Variance framework in `07_ML_Theory/`).

**SARSA** (on-policy TD control): same idea, applied to action-values:
$$Q(s_t,a_t)\leftarrow Q(s_t,a_t)+\alpha\big[R_t+\gamma Q(s_{t+1},a_{t+1})-Q(s_t,a_t)\big]$$
