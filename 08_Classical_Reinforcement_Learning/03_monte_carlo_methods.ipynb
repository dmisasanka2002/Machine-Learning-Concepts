Estimate $V^\pi(s)$ by averaging **observed** returns from sampled episodes
(no need to know $P$ — this is the key departure from Dynamic Programming,
important to state explicitly: DP requires a known model, Monte Carlo and
TD do not).
$$V^\pi(s) \approx \frac{1}{N}\sum_{i=1}^N G_i(s), \qquad G_i(s)=\text{observed discounted return from episode } i \text{ starting at } s$$
