# **ML Workflow**

**Prerequisites:** Learning Paradigms | **Next:** Problem Formulation | **Depth tier:** Foundation

## 1. Theory

A repeatable end-to-end process, so later sections of this repo (Data →
Supervised/Unsupervised → Evaluation → Optimization) map onto concrete
pipeline stages rather than feeling like an arbitrary folder order.

## 2. The Stages (and where each is covered in this repo)

1. **Problem formulation** — `01_Foundations/05_problem_formulation.ipynb`
2. **Data collection & exploration** — `02_Data/01_data_types.ipynb`, `02_sampling.ipynb`
3. **Data cleaning & preprocessing** — `02_Data/03-06`
4. **Feature engineering** — `02_Data/08_feature_engineering.ipynb`
5. **Train/validation/test split** — `05_Model_Evaluation/05_cross_validation.ipynb`, `07_validation_strategies.ipynb`
6. **Model selection & training** — `03_Supervised_Learning/`, `04_Unsupervised_Learning/`
7. **Hyperparameter tuning** — `06_Optimization/07_hyperparameter_optimization.ipynb`
8. **Evaluation** — `05_Model_Evaluation/`
9. **Interpretation** — `09_Interpretability/`
10. **Deployment & monitoring** — out of scope for this repo (an engineering topic more than an ML-concepts one; flagged rather than built out, per the same scope discipline as semi-supervised learning above)
