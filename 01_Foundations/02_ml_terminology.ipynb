## 2. Core Vocabulary (grouped, not alphabetical - grouping aids retention)

**Data**: instance/sample/example (one row), feature/attribute/predictor (one column of $X$), label/target/response ($y$), feature vector ($x_i\in\mathbb R^d$), design matrix ($X\in\mathbb R^{n\times d}$).

**Model**: parameters (learned from data, e.g. $w$ in Linear Regression) vs. hyperparameters (set before training, e.g. $k$ in KNN, $\lambda$ in Ridge — the entire subject of `06_Optimization/` and hyperparameter
search). Hypothesis / model / estimator - near-synonyms in casual use; `07_ML_Theory/01_hypothesis_spaces_and_erm.ipynb` uses "hypothesis" precisely (one candidate function from $\mathcal H$).

**Training**: loss/cost/objective function (per-example vs. aggregate - worth distinguishing, since "loss" often means per-example and "cost" the sum/average, though usage varies by author), epoch (one full pass over training data, relevant once Gradient Descent moves to mini-batch/SGD), convergence.

**Evaluation**: train/validation/test split - three different roles: fit parameters, tune hyperparameters, final unbiased estimate — a common point of confusion worth stating precisely (using test data to tune
hyperparameters is exactly the leakage `02_Data/07_data_leakage.ipynb` warns about).
